```python
#Instalar los programas necesarios
!wget -q https://dl.google.com/linux/direct/google-chrome-stable_current_amd64.deb
!apt-get install -y ./google-chrome-stable_current_amd64.deb
!pip install selenium webdriver-manager

#Cargar librerias
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
from IPython.display import Image, display
import time
```

```python
class TestCarrito():
  def __init__(self):
    # Configuración de Chrome
    self.options = Options()
    self.options.add_argument("--headless")
    self.options.add_argument("--no-sandbox")
    self.options.add_argument("--disable-dev-shm-usage")
    # Esto establece un tamaño muy grande, forzando una captura más completa.
    self.options.add_argument("--window-size=1920,1080")
    # Crear web driver
    self.service = Service(ChromeDriverManager().install())


  def ejecutarTestBasico(self):
    #crear driver
    driver = webdriver.Chrome(service=self.service, options=self.options)

    # ir a la pagina web
    driver.get("https://www.demoblaze.com/index.html")
    time.sleep(1)
    driver.save_screenshot("Paso1.png")

    #Tocamos link a laptops
    botonLaptops = driver.find_element(By.LINK_TEXT, "Laptops")
    botonLaptops.click()
    time.sleep(1)
    driver.save_screenshot("Paso2.png")

    #seleccionamos la notebook que queremos comprar
    imagenLaptops = driver.find_element(By.XPATH, '//*[@id="tbodyid"]/div[3]/div/a/img')
    imagenLaptops.click()
    time.sleep(1)
    driver.save_screenshot("Paso3.png")

    #Agregar al carrito
    botonCarrito = driver.find_element(By.CLASS_NAME, "btn-success")
    botonCarrito.click()
    time.sleep(1)

    #Aceptar alerta
    alerta = driver.switch_to.alert
    alerta.accept()

    #ir al carrito
    linkCarrito = driver.find_element(By.ID, "cartur")
    linkCarrito.click()
    time.sleep(1)
    driver.save_screenshot("Paso4.png")

    #comprar
    botonComprar = driver.find_element(By.XPATH, '//*[@id="page-wrapper"]/div/div[2]/button')
    botonComprar.click()
    time.sleep(1)
    driver.save_screenshot("Paso5.png")

    #Rellenar formulario
    campoTexto1 = driver.find_element(By.ID, "name")
    campoTexto1.send_keys("Roberto")

    campoTexto2 = driver.find_element(By.ID, "country")
    campoTexto2.send_keys("Roberto")

    campoTexto3 = driver.find_element(By.ID, "city")
    campoTexto3.send_keys("Roberto")

    campoTexto4 = driver.find_element(By.ID, "card")
    campoTexto4.send_keys("Roberto")

    campoTexto5 = driver.find_element(By.ID, "month")
    campoTexto5.send_keys("Roberto")

    campoTexto6 = driver.find_element(By.ID, "year")
    campoTexto6.send_keys("Roberto")

    driver.save_screenshot("Paso6.png")

    #Confirmar compra
    botonConfirmar = driver.find_element(By.XPATH, '//*[@id="orderModal"]/div/div/div[3]/button[2]')
    botonConfirmar.click()
    time.sleep(1)
    driver.save_screenshot("Paso7.png")




    #mostrar las imagenes
    display(Image("Paso1.png"))
    display(Image("Paso2.png"))
    display(Image("Paso3.png"))
    display(Image("Paso4.png"))
    display(Image("Paso5.png"))
    display(Image("Paso6.png"))
    display(Image("Paso7.png"))


    #MUCHO MUY IMPORTANTE
    driver.quit()
```

```python
test = TestCarrito()
test.ejecutarTestBasico()
```
